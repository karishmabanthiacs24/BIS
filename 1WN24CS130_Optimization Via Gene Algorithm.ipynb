```python
import random


def objective_function(x):
    return x**2 + 4*x + 4



POPULATION_SIZE = 5
MUTATION_RATE = 0.1
CROSSOVER_RATE = 0.8
GENERATIONS = 20



population = [random.uniform(-10, 10) for _ in range(POPULATION_SIZE)]



for generation in range(GENERATIONS):


    best_x = min(population, key=objective_function)
    best_value = objective_function(best_x)


    if generation % 5 == 0:
        print("Generation:", generation)
        print("Best x:", round(best_x, 4))
        print("Best fitness value:", round(best_value, 4))
        print("------------------------")

   
    new_population = []

    while len(new_population) < POPULATION_SIZE:

    
        parent1 = min(random.sample(population, 2),
                      key=objective_function)

        parent2 = min(random.sample(population, 2),
                      key=objective_function)

  
        if random.random() < CROSSOVER_RATE:
            child = (parent1 + parent2) / 2
        else:
            child = parent1

     
        if random.random() < MUTATION_RATE:
            child += random.uniform(-1, 1)

        new_population.append(child)


    population = new_population



best_x = min(population, key=objective_function)
best_value = objective_function(best_x)

print("\n========== FINAL RESULT ==========")
print("Best solution (x) =", round(best_x, 4))
print("Minimum value f(x) =", round(best_value, 4))
print("==================================")
print("Karishma Banthia(1WN24CS130)")
```
