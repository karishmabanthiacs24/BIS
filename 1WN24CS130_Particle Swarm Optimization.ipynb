```python
import random

particles = [10, 15, 30, 40, 20]
velocity = [0] * 5
pbest = particles.copy()

target = 25
gbest = min(pbest, key=lambda x: abs(x - target))

for i in range(10):
    for j in range(5):

        r1 = random.random()
        r2 = random.random()

        velocity[j] = (0.7 * velocity[j] +
                       1.5 * r1 * (pbest[j] - particles[j]) +
                       1.5 * r2 * (gbest - particles[j]))

        particles[j] += velocity[j]

        if abs(particles[j] - target) < abs(pbest[j] - target):
            pbest[j] = particles[j]

        if abs(pbest[j] - target) < abs(gbest - target):
            gbest = pbest[j]

    print("Iteration", i + 1, "Best Temperature:", round(gbest, 2))

print("\nTarget Temperature:", target)
print("Optimal Temperature:", round(gbest, 2))
print("Karishma Banthia(1WN24CS130)")
```
